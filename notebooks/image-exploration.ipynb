# Simple Notebook for Image Processing

```python
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
DATA_DIR = Path("./../data")
IMAGE_PATH = DATA_DIR / "development_set/3f_png.rf.81c7e132365ef95bb19380ca389025f6.jpg"
```

```python
image = cv2.imread(str(IMAGE_PATH))
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display the image
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
image = cv2.GaussianBlur(image, (15, 15), 0)
plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
hsv_image = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)

lower_blue = np.array([200/2, 60 / 100 * 255, 60 / 100 * 255])
upper_blue = np.array([210/2, 90 / 100 * 255, 90 / 100 * 255])

mask_table = cv2.inRange(hsv_image, lower_blue, upper_blue)

plt.imshow(mask_table, cmap='gray')
plt.axis('off')
plt.title('Mask for Blue Color')
plt.show()

# Plot all three channels of the HSV image
fig, axes = plt.subplots(1, 3, figsize=(12, 5))
axes[0].imshow(hsv_image[:, :, 0], cmap='hsv')
axes[0].set_title('H')
axes[0].axis('off')
axes[1].imshow(hsv_image[:, :, 1], cmap='gray')
axes[1].set_title('S')
axes[1].axis('off')
axes[2].imshow(hsv_image[:, :, 2], cmap='gray')
axes[2].set_title('V')
axes[2].axis('off')
plt.show()
```
